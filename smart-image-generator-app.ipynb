```python
import gradio as gr
import torch
from diffusers import StableDiffusionPipeline, StableDiffusionXLPipeline
from PIL import Image
```

```python
# Preload both models
print("Loading SD1.5...")
pipe_sd15 = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5",
    torch_dtype=torch.float16
).to("cuda")

print("Loading SDXL...")
pipe_sdxl = StableDiffusionXLPipeline.from_pretrained(
    "stabilityai/stable-diffusion-xl-base-1.0",
    torch_dtype=torch.float16,
    variant="fp16"
).to("cuda")
```

```python
# Image generation function
from gradio import update
def generate_image(prompt, model_name, width, height, cfg):
    if not prompt.strip():
        return None, update(value="⚠️ Please enter a prompt before generating an image.", visible=True)

    pipe = pipe_sd15 if model_name == "SD 1.5" else pipe_sdxl
    with torch.inference_mode():
        image = pipe(prompt, height=height, width=width, guidance_scale=cfg).images[0]

    return image, update(value="", visible=False)

# Example prompts
examples = [
    ["a futuristic cityscape at sunset"],
    ["a cyberpunk cat with sunglasses"],
    ["a magical forest with glowing trees"],
    ["an astronaut riding a horse on Mars"],
    ["a portrait of a lion in royal armor"]
]

# App UI
with gr.Blocks(theme=gr.themes.Soft(), css="""
.gradio-container {
    background: linear-gradient(135deg, #4F46E5, #800020);
    min-height: 100vh;
    margin: 0;
    padding: 20px;
}

/* Button */
#generate-btn {
    background-color: #4F46E5;
    color: white;
    font-size: 20px;
    font-weight: bold;
    border-radius: 12px;
    padding: 14px 28px;
    border: none;
    margin: 0 auto;
    display: block;
    width: fit-content;
    transition: background 0.3s ease;
}
#generate-btn:hover {
    background-color: #800020;
    cursor: pointer;
}

/* Error */
#error-msg {
    color: red;
    font-weight: bold;
    text-align: center;
    margin-top: 10px;
}

/* Slider Gradient Track */
input[type="range"]::-webkit-slider-runnable-track {
    background: linear-gradient(135deg, #4F46E5, #800020);
    height: 6px;
    border-radius: 4px;
}
input[type="range"]::-moz-range-track {
    background: linear-gradient(135deg, #4F46E5, #800020);
    height: 6px;
    border-radius: 4px;
}

/* Slider Thumb */
input[type="range"]::-webkit-slider-thumb {
    background-color: white;
    border: 2px solid #800020;
    border-radius: 50%;
    height: 20px;
    width: 20px;
    margin-top: -7px;
    cursor: pointer;
    appearance: none;
}
input[type="range"]::-moz-range-thumb {
    background-color: white;
    border: 2px solid #800020;
    border-radius: 50%;
    height: 20px;
    width: 20px;
    cursor: pointer;
}
input[type="radio"]:checked {
    background-color: #800020 !important;
    border-color: #800020 !important;
    color: white !important;
}
""") as demo:

    gr.Markdown("# 🎨 Smart AI Image Generator")
    gr.Markdown("### Choose model and type your prompt below to generate art.")

    with gr.Row():
        prompt = gr.Textbox(label="Prompt", placeholder="A neon tiger in space")
        model_choice = gr.Radio(["SD 1.5", "SDXL"], value="SD 1.5", label="Choose Model")

    gr.Examples(
        examples=examples,
        inputs=[prompt],
        label="Try an example prompt"
    )

    gr.Markdown("### Choose image size and CFG.")

    with gr.Row():
        width = gr.Slider(512, 1024, step=64, value=640, label="Image Width")
        height = gr.Slider(512, 1024, step=64, value=640, label="Image Height")
        cfg = gr.Slider(1.0, 20.0, step=0.5, value=7.5, label="CFG (Guidance Scale)")

    generate_button = gr.Button("Generate Image", elem_id="generate-btn")
    output_image = gr.Image(label="Generated Image")
    error_msg = gr.Label(value="Error", elem_id="error-msg", visible=False)


    # When button is clicked, update image and error message
    generate_button.click(
        fn=generate_image,
        inputs=[prompt, model_choice, width, height, cfg],
        outputs=[output_image, error_msg]
    )

demo.launch(share=True)
```
